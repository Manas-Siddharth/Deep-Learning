```python
import pandas as pd
import numpy as np
data = pd.read_csv('housing.csv', header=0)
data.head()
data.shape
```

```python
import pandas as pd
import numpy as np
from sklearn import preprocessing

x = data.drop('median_house_value', axis=1)
x = x.select_dtypes(include=np.number).copy()
x.fillna(x.mean(), inplace=True)
min_max_scaler = preprocessing.MinMaxScaler()
x_scale = min_max_scaler.fit_transform(x)
print(x_scale)
```

```python
from sklearn.model_selection import train_test_split
from sklearn import preprocessing

y_unscaled= data['median_house_value']
min_max_y = preprocessing.MinMaxScaler()
y_scale = min_max_y.fit_transform(y_unscaled.values.reshape(-1, 1))
x_train, x_test, y_train, y_test = train_test_split(x_scale,y_scale,test_size=0.30,random_state=0)
```

```python
from tensorflow import keras
from keras.models import Sequential

model = Sequential()
model.add(keras.Input(shape=(x_train.shape[1],)))
model.add(keras.layers.Dense(128,activation='relu'))
model.add(keras.layers.Dense(64,activation='relu'))
model.add(keras.layers.Dropout(0.3))
model.add(keras.layers.Dense(1,activation='linear'))

model.compile(optimizer='adam',loss='mean_squared_error',metrics=['mean_absolute_error'])

model.fit(x_train,y_train,epochs=10,batch_size=64)
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error

mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
print("MSE:", mse)
print("MAE:", mae)
```
